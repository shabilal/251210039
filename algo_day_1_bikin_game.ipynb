```python
from IPython.display import HTML

HTML("""
<canvas id="spaceGame" width="400" height="500"
        style="border:1px solid #000; background:#000;"></canvas>
<p>Kontrol: <b>←</b> / <b>→</b>, <b>Space</b> untuk menembak.</p>

<script>
  const canvas = document.getElementById('spaceGame');
  const ctx = canvas.getContext('2d');

  // Pesawat
  const shipWidth = 40;
  const shipHeight = 20;
  let shipX = (canvas.width - shipWidth) / 2;
  const shipY = canvas.height - shipHeight - 10;
  const shipSpeed = 5;

  // Kontrol
  let leftPressed = false;
  let rightPressed = false;
  let spacePressed = false;
  let canShoot = true;

  // Peluru
  const bullets = [];
  const bulletWidth = 4;
  const bulletHeight = 10;
  const bulletSpeed = 7;

  // Musuh
  const enemies = [];
  const enemyWidth = 30;
  const enemyHeight = 20;
  let enemySpeed = 2;
  let enemySpawnInterval = 80; // tiap 80 frame spawn musuh baru

  let frameCount = 0;
  document.addEventListener('keydown', e => {
    if (e.key === 'ArrowLeft' || e.key === 'Left')  leftPressed = true;
    if (e.key === 'ArrowRight' || e.key === 'Right') rightPressed = true;
    if (e.code === 'Space') spacePressed = true;
  });

  document.addEventListener('keyup', e => {
    if (e.key === 'ArrowLeft' || e.key === 'Left')  leftPressed = false;
    if (e.key === 'ArrowRight' || e.key === 'Right') rightPressed = false;
    if (e.code === 'Space') { spacePressed = false; canShoot = true; }
  });

  function drawShip() {
    ctx.fillStyle = '#00FF00';
    ctx.fillRect(shipX, shipY, shipWidth, shipHeight);
    ctx.fillStyle = '#00AA00';
    ctx.fillRect(shipX + shipWidth/2 - 5, shipY - 5, 10, 5);
  }
  function updateShip() {
    if (leftPressed && shipX > 0) shipX -= shipSpeed;
    if (rightPressed && shipX < canvas.width - shipWidth) shipX += shipSpeed;

    if (spacePressed && canShoot) {
      bullets.push({ x: shipX + shipWidth/2 - bulletWidth/2, y: shipY - bulletHeight });
      canShoot = false;
    }
  }

  function updateBullets() {
    bullets.forEach(b => b.y -= bulletSpeed);
    for (let i = bullets.length - 1; i >= 0; i--) {
      if (bullets[i].y + bulletHeight < 0) bullets.splice(i, 1);
    }
  }

  function drawBullets() {
    ctx.fillStyle = '#FFFF00';
    bullets.forEach(b => ctx.fillRect(b.x, b.y, bulletWidth, bulletHeight));
  }
  function spawnEnemy() {
    const x = Math.random() * (canvas.width - enemyWidth);
    enemies.push({ x: x, y: -enemyHeight });
  }

  function updateEnemies() {
    enemies.forEach(en => en.y += enemySpeed);
    // Hapus yang keluar layar
    for (let i = enemies.length - 1; i >= 0; i--) {
      if (enemies[i].y > canvas.height) {
        enemies.splice(i, 1);
      }
    }
  }

  function drawEnemies() {
    ctx.fillStyle = '#FF0000';
    enemies.forEach(en => ctx.fillRect(en.x, en.y, enemyWidth, enemyHeight));
  }

  function drawBackground() {
    ctx.fillStyle = '#000000';
    ctx.fillRect(0, 0, canvas.width, canvas.height);
  }

  function gameLoop() {
    frameCount++;

    drawBackground();
    updateShip();
    updateBullets();
    updateEnemies();

    if (frameCount % enemySpawnInterval === 0) {
      spawnEnemy();
    }

    drawShip();
    drawBullets();
    drawEnemies();

    requestAnimationFrame(gameLoop);
  }

  gameLoop();
</script>
""")
```

```python
from IPython.display import HTML

HTML("""
<!-- Bagian HTML: judul, info kontrol, canvas, dan tampilan skor/nyawa -->
<div style="font-family: Arial; max-width: 420px;">
  <h3>🚀 Game Pesawat Menembak</h3>
  <p>Kontrol: <b>←</b> / <b>→</b> untuk gerak, <b>Space</b> untuk menembak.</p>
  <!-- Canvas tempat game digambar -->
  <canvas id="spaceGame" width="400" height="500" style="border:1px solid #000; background:#000;"></canvas>
  <!-- Tempat menampilkan skor dan nyawa (akan di-update dari JavaScript) -->
  <p>Skor: <span id="score">0</span> | Nyawa: <span id="lives">3</span></p>
</div>

<script>
  // Ambil elemen canvas dan context 2D untuk menggambar
  const canvas = document.getElementById('spaceGame');
  const ctx = canvas.getContext('2d');
  // =========================
  // 1. VARIABEL PESAWAT
  // =========================
  const shipWidth = 40;              // lebar pesawat
  const shipHeight = 20;             // tinggi pesawat
  let shipX = (canvas.width - shipWidth) / 2; // posisi X awal (tengah)
  const shipY = canvas.height - shipHeight - 10; // posisi Y tetap di dekat bawah
  const shipSpeed = 5;               // kecepatan gerak pesawat

  // =========================
  // 2. VARIABEL KONTROL / INPUT
  // =========================
  let leftPressed = false;           // apakah tombol kiri ditekan?
  let rightPressed = false;          // apakah tombol kanan ditekan?
  let spacePressed = false;          // apakah tombol spasi ditekan?
  let canShoot = true;               // untuk mencegah spam peluru (1 peluru per tekan spasi)
  // =========================
  // 3. VARIABEL PELURU
  // =========================
  const bullets = [];                // array untuk menyimpan semua peluru aktif
  const bulletWidth = 4;             // lebar peluru
  const bulletHeight = 10;           // tinggi peluru
  const bulletSpeed = 7;             // kecepatan peluru naik ke atas

  // =========================
  // 4. VARIABEL MUSUH
  // =========================
  const enemies = [];                // array untuk menyimpan semua musuh aktif
  const enemyWidth = 30;             // lebar musuh
  const enemyHeight = 20;            // tinggi musuh
  let enemySpeed = 2;                // kecepatan musuh turun
  let enemySpawnInterval = 80;       // tiap berapa frame musuh baru muncul (semakin kecil = makin sering)
  // =========================
  // 5. VARIABEL GAME STATE
  // =========================
  let frameCount = 0;                // menghitung berapa banyak frame yang sudah lewat
  let score = 0;                     // skor pemain
  let lives = 3;                     // jumlah nyawa
  let gameOver = false;              // apakah game sudah berakhir?

  // Tampilkan skor dan nyawa awal di HTML
  document.getElementById('score').textContent = score;
  document.getElementById('lives').textContent = lives;

  // =========================
  // 6. EVENT LISTENER KEYBOARD
  // =========================
  // Dengarkan event keyboard untuk input pemain
  document.addEventListener('keydown', keyDownHandler);
  document.addEventListener('keyup', keyUpHandler);
  function keyDownHandler(e) {
    // Tombol kiri
    if (e.key === 'ArrowLeft' || e.key === 'Left') {
      leftPressed = true;
    // Tombol kanan
    } else if (e.key === 'ArrowRight' || e.key === 'Right') {
      rightPressed = true;
    // Tombol spasi
    } else if (e.code === 'Space') {
      spacePressed = true;
    }
  }

  function keyUpHandler(e) {
    // Lepas tombol kiri
    if (e.key === 'ArrowLeft' || e.key === 'Left') {
      leftPressed = false;
    // Lepas tombol kanan
    } else if (e.key === 'ArrowRight' || e.key === 'Right') {
      rightPressed = false;
    // Lepas tombol spasi
    } else if (e.code === 'Space') {
      spacePressed = false;
      canShoot = true; // setelah spasi dilepas, boleh menembak lagi
    }
  }
  // =========================
  // 7. GAMBAR PESAWAT
  // =========================
  function drawShip() {
    // Badan pesawat
    ctx.fillStyle = '#00FF00';
    ctx.fillRect(shipX, shipY, shipWidth, shipHeight);
    // Cockpit kecil di atas
    ctx.fillStyle = '#00AA00';
    ctx.fillRect(shipX + shipWidth/2 - 5, shipY - 5, 10, 5);
  }

  // =========================
  // 8. UPDATE POSISI PESAWAT & TEMBAKAN
  // =========================
  function updateShip() {
    // Gerak kiri kalau tombol kiri ditekan dan belum menyentuh batas kiri
    if (leftPressed && shipX > 0) {
      shipX -= shipSpeed;
    }
    // Gerak kanan kalau tombol kanan ditekan dan belum menyentuh batas kanan
    if (rightPressed && shipX < canvas.width - shipWidth) {
      shipX += shipSpeed;
    }
    // Tembak peluru ketika spasi ditekan, boleh menembak, dan game belum game over
    if (spacePressed && canShoot && !gameOver) {
      bullets.push({
        x: shipX + shipWidth/2 - bulletWidth/2, // posisi X di tengah pesawat
        y: shipY - bulletHeight                 // posisi Y di atas pesawat
      });
      canShoot = false; // peluru baru hanya boleh setelah spasi dilepas
    }
  }
  // =========================
  // 9. UPDATE & GAMBAR PELURU
  // =========================
  function updateBullets() {
    // Naikkan posisi Y semua peluru (ke atas)
    bullets.forEach(b => b.y -= bulletSpeed);

    // Hapus peluru yang sudah keluar dari atas layar
    for (let i = bullets.length - 1; i >= 0; i--) {
      if (bullets[i].y + bulletHeight < 0) {
        bullets.splice(i, 1); // hapus 1 elemen di index i
      }
    }
  }

  function drawBullets() {
    ctx.fillStyle = '#FFFF00'; // warna peluru kuning
    bullets.forEach(b => ctx.fillRect(b.x, b.y, bulletWidth, bulletHeight));
  }
  // =========================
  // 10. MUSUH: SPAWN, UPDATE, GAMBAR
  // =========================
  // Buat musuh baru di posisi X acak di atas layar
  function spawnEnemy() {
    const x = Math.random() * (canvas.width - enemyWidth); // acak antara 0 sampai (lebar canvas - lebar musuh)
    enemies.push({ x: x, y: -enemyHeight }); // y negatif agar muncul dari atas
  }

  function updateEnemies() {
    // Turunkan semua musuh
    enemies.forEach(en => en.y += enemySpeed);

    // Cek musuh yang keluar layar bawah → kurangi nyawa
    for (let i = enemies.length - 1; i >= 0; i--) {
      if (enemies[i].y > canvas.height) {
        enemies.splice(i, 1);    // hapus musuh
        lives--;                 // nyawa berkurang
        document.getElementById('lives').textContent = lives; // update di HTML

        // Kalau nyawa habis, akhiri game
        if (lives <= 0) {
          endGame();
        }
      }
    }
  }
  function drawEnemies() {
    ctx.fillStyle = '#FF0000'; // warna musuh merah
    enemies.forEach(en => ctx.fillRect(en.x, en.y, enemyWidth, enemyHeight));
  }

  // =========================
  // 11. CEK TABRAKAN (COLLISION)
  // =========================
  function checkCollisions() {
    // -------- Peluru vs Musuh --------
    for (let i = enemies.length - 1; i >= 0; i--) {
      for (let j = bullets.length - 1; j >= 0; j--) {
        const en = enemies[i];
        const b = bullets[j];

        // Cek overlap kotak peluru dan kotak musuh (collision detection sederhana berbasis bounding box)
        if (
          b.x < en.x + enemyWidth &&
          b.x + bulletWidth > en.x &&
          b.y < en.y + enemyHeight &&
          b.y + bulletHeight > en.y
        ) {
          // Kalau bertabrakan: hapus musuh dan peluru
          enemies.splice(i, 1);
          bullets.splice(j, 1);
          // Tambah skor
          score++;
          document.getElementById('score').textContent = score;

          // Setiap kelipatan 5 skor → tingkatkan kesulitan
          if (score % 5 === 0) {
            enemySpeed += 0.3;         // musuh makin cepat
            if (enemySpawnInterval > 30) { // jangan terlalu sering sekali
              enemySpawnInterval -= 5; // musuh spawn lebih sering
            }
          }
          break; // keluar dari loop peluru, karena musuh sudah dihapus
        }
      }
    }
    // -------- Musuh vs Pesawat --------
    for (let i = enemies.length - 1; i >= 0; i--) {
      const en = enemies[i];
      // Cek overlap kotak musuh dan kotak pesawat
      if (
        en.x < shipX + shipWidth &&
        en.x + enemyWidth > shipX &&
        en.y < shipY + shipHeight &&
        en.y + enemyHeight > shipY
      ) {
        // Kalau musuh menabrak pesawat: hapus musuh & kurangi nyawa
        enemies.splice(i, 1);
        lives--;
        document.getElementById('lives').textContent = lives;

        if (lives <= 0) {
          endGame();
        }
      }
    }
  }
  // =========================
  // 12. BACKGROUND BINTANG
  // =========================
  function drawBackgroundStars() {
    // Latar belakang hitam
    ctx.fillStyle = '#000';
    ctx.fillRect(0, 0, canvas.width, canvas.height);

    // Bintik-bintik abu-abu sebagai bintang acak
    ctx.fillStyle = '#444';
    for (let i = 0; i < 40; i++) {
      const x = Math.random() * canvas.width;
      const y = Math.random() * canvas.height;
      ctx.fillRect(x, y, 1, 1);
    }
  }

  // =========================
  // 13. AKHIR GAME
  // =========================
  function endGame() {
    gameOver = true; // berhenti update di gameLoop
    alert("Game Over! Skor kamu: " + score); // popup sederhana
  }
  // =========================
  // 14. GAME LOOP UTAMA
  // =========================
  function gameLoop() {
    // Hanya jalan kalau game belum berakhir
    if (!gameOver) {
      frameCount++; // tambah 1 setiap frame

      // Gambar background
      drawBackgroundStars();

      // Update logika
      updateShip();
      updateBullets();
      updateEnemies();
      checkCollisions();

      // Spawn musuh berkala berdasarkan frameCount dan interval
      if (frameCount % enemySpawnInterval === 0) {
        spawnEnemy();
      }

      // Gambar semua objek
      drawBullets();
      drawEnemies();
      drawShip();
      // Minta frame berikutnya (animasi)
      requestAnimationFrame(gameLoop);
    } else {
      // Kalau game sudah over, gelapkan layar dan tulis pesan
      ctx.fillStyle = 'rgba(0, 0, 0, 0.6)';
      ctx.fillRect(0, 0, canvas.width, canvas.height);
      ctx.fillStyle = '#FFFFFF';
      ctx.font = '24px Arial';
      ctx.fillText('GAME OVER', canvas.width / 2 - 70, canvas.height / 2);
      ctx.font = '16px Arial';
      ctx.fillText('Jalankan ulang sel untuk main lagi', canvas.width / 2 - 120, canvas.height / 2 + 30);
    }
  }

  // Mulai game pertama kali
  gameLoop();
</script>
""")
```
